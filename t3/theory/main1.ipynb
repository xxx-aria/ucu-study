match-case

```python
num = 13
match num:
    case 1:
        print('one')
    case 2 | 3 | 4 | 5:
        print('too much')
    case num if num < 10: # bad practice
        print(num)
    case x if x in (11, 12, 13): # x = num
        print(f'in tuple - {x}')
    case x if 13 < x < 20: # x = num, нова змінна автоматично
        print(x)
    case _: # case wildcard
        print('default case')
```

for-else

```python
for i in range(5):
    if i == 3:
        continue
    print(i, end=' ')
else:
    print('the end')
```

```python
for i in range(5):
    print(i, end=' ')
    if i == 3:
        break
else:
    print('the end')
```
