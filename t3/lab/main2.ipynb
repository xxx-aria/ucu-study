### if-elif-else and match-case

```python
grade = 8
match grade:
    case low_grade if 0 < low_grade <= 7:
        print(f'low grade, {grade}')
    case good_grade if good_grade <= 10:
        print(f'good grade, {grade}')
    case great_grade if great_grade <= 12:
        print(f'great grade, {grade}')
    case _:
        print(f'hm {grade}')
```

```python
status = 1403

match status:
    case 200:
        print('OK')
    case stat if stat == 400: # stat = status
        print('Bad request')
    case 401 | 403 | 404:
        print('Not allowed')
    case st if st == 418 or st > 1000: # st = status
        print('I\'m a teapot')
    case _:
        print('Something\'s wrong with the internet')
```

```python
point = (1, 0)
match point:
    case (0, 0):
        print('origin')
    case (0, y): # (0, point[1])
        print(f'{y = }')
    case (x, 0): # (point[0], 0)
        print(f'{x = }')
    case (x, y): # point
        print(f'{x = }, {y = }')
    case _:
        print('not a point')
```

```python
point = (1, 0)
match point:
    case (0, y) if y == 1 / 0: # skipped for (1, 0)
        print(f'oops!')
    case (x, y):
        print('point')
```
