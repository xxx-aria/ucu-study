formatting strings

```python
num = 1
print('num is', num)
# print('num is ' + num) # TypeError
print('num is' + str(num))
print(f'num is {num}')
```

```python
num = 3.1415
print(f'num is {num:.2f}')
print('num is %.2f'%(num))
```

```python
num = 3.15
print(f'{num:.1f}')
num = 3.16
print(f'{num:.1f}')
num = 3.5
print(f'{num:.0f}')
num = 4.5
print(f'{num:.0f}')
```

```python
num = 1.00002
print(f'{num:.2f}') # ці 2 цифри будуть завжди
print(round(num, 2))
```

:= моржовий оператор, моржеве присвоєння
- inside an expression (лише всередині виразу)

```python
n1 = 10
n2 = 5
# присвоїти і повернути
print(res := n1 + n2)
# res := n1 + n2 # SyntaxError, бо просто так не можна присвоїти
```

```python
n1 = 1
n2 = 2
(res := n1 + n2) > 2
```

```python
n1 = 1
n2 = 2
# print(f'{res:=n1+n2:.2f}') # Error
res = 0
print(f'{1 + (res := 1 + 2)}')
```
