```python
def merge_sort(lst):
    if len(lst) <= 1:
        return lst
    mid = len(lst) // 2
    left = merge_sort(lst[:mid])
    right = merge_sort(lst[mid:])
    res = []
    i, j = 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            res.append(left[i])
            i += 1
        else:
            res.append(right[j])
            j += 1
    if i < len(left):
        res.extend(left[i:])
    if j < len(right):
        res.extend(right[j:])
    return res
```

```python
def quick_sort(lst):
    lst = lst.copy()

    if len(lst) <= 1:
        return lst

    pivot = len(lst) // 2
    left, right = [], []
    for i in range(len(lst)):
        if i == pivot:
            continue
        if lst[i] < lst[pivot]:
            left.append(lst[i])
        else:
            right.append(lst[i])

    left = quick_sort(left)
    right = quick_sort(right)

    return left + [lst[pivot]] + right
```
