```python
lst = [9, 13, 21]
n = 14
for num in lst:
    if num > n:
        print('n')
        break
    if num == n:
        print('y')
        break
```

```python
def bin_search(n):
    l, r = 0, len(lst)
    while l<=r:
        mid = (l + r + 1)//2

        if lst[mid] == n:
            print(mid)
            return mid
            break

        if lst[mid] < n:
            l = mid
        elif lst[mid] > n:
            r = mid


lst = [2, 7, 11, 15, 24, 26, 37, 40, 49, 51, 55]

[bin_search(x) == lst.index(x) for x in [2, 49, 55]]
```
