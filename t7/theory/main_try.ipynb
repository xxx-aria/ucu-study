flush in print
- flushes buffered stuff and prints it immedietely 

sys.stdout.flush_interval - default it 0.2

```python
import sys, time

sys.stdout.flush_interval = 10   # auto-flush only every 10 s now
```

```python
for el in 'python':
    print(el, end=' ', flush=False)
    time.sleep(1)
```

```python
for el in 'python':
    print(el, end=' ', flush=True)
    time.sleep(1)
```

```python
sys.stdout.flush_interval = 0.2
```

#### files

```python
f = open('myfile.txt', encoding='utf-8')
print(f)
f.close()
```

```python
f = open('myfile.txt', encoding='utf-8')
print(f.closed)
try:
    5 / 0
    f.close()
finally:
    print(f.closed)
```

```python
with open('myfile.txt', encoding='utf-8') as f:
    print(f.readable())
    print(f.writable())
```

```python
with open('myfile.txt', 'r', encoding='utf-8') as f:
    text = f.read()
    print(text)
    print(repr(text))
```

```python
with open('myfile.txt', 'r', encoding='utf-8') as f:
    print(repr(f.readline()))
    print(repr(f.readline()))
```

```python
with open('myfile.txt', 'r', encoding='utf-8') as f:
    print(f.tell())
    print(repr(f.readline()))
    print(f.tell())
    print(repr(f.readline()))
    print(f.seek(1))
    print(f.read())
    print(f.tell())
```

```python
with open('myfile.txt', 'r', encoding='utf-8') as f:
    f.seek(13)
    print(f.read(5))
```

```python
with open('myfile.txt', 'r', encoding='utf-8') as f:
    for line in f:
        print(line)
        print(repr(line))
        print(repr(line.strip('\n')))
```

```python
with open('myfile_w.txt', 'w', encoding='utf-8') as f:
    content = [1, 2, 3]
    f.write('\n'.join(str(x) for x in content))
```
