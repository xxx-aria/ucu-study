```python
fs = frozenset({1, 2})
h1 = hash(fs)
fs |= {3}
h2 = hash(fs)
h1 == h2
```

```python
t = (1, 2)
h1 = hash(t)
t += (3,)
h2 = hash(t)
h1 == h2
```

```python
def ct(L) -> tuple[set, set, set]:
    L += L
    M = [(el + i) for i, el in enumerate(L)]
    s = set(L)
    t = set(M)
    print(s, t)
    return s-t, t|s, s&t
print(ct([2, 4, 8]))
```
