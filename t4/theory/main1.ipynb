### try-except statements

```python
a, b = 1, 0
try:
    print('start division')
    a / b
    print('end division')
except ZeroDivisionError:
    print('denom is 0')
```

```python
a, b = 'hello', 0
try:
    print('start division')
    a / b
    print('end division')
except ZeroDivisionError:
    print('zero division error : denom is 0')
except TypeError:
    print('type error : division of unsupported ops')
```

```python
a, b = 'hello', 0
try:
    print('start division')
    a / b
    print('end division')
except ZeroDivisionError, TypeError, SyntaxError: # can be in ( )
    print('oops')
```

```python
a, b = 'hello', 0
```

```python
try:
    print('start division')
    a / b
    print('end division')
except (ZeroDivisionError, TypeError) as e:
    print('oops', type(e), e)
```
