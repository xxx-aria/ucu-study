#### list comprehension

```python
res_1 = []
for i in range(5):
    res_1.append(i)

res_2 = [i for i in range(5)]

res_1 == res_2
```

```python
[print(1) for i in range(3)]
```

```python
[(print(1), i) for i in range(3)]
```

```python
generator = (i ** 2 for i in range(5))
print(generator)
tuple(generator) # generates one element at once
```

```python
tuple([i ** 2 for i in range(5)]) # memory overuse, generates whole list and then goes one by one
```

#### mutable and immutable

```python
a = (1, 2)
id_1 = id(a)
a += (3, 4)
id_2 = id(a)
a, id_1 == id_2
```

```python
a = [1, 2]
id_1 = id(a)
a += [3, 4]
id_2 = id(a)
a, id_1 == id_2
```

```python
a = [1, 2]
a_2 = a
a_copy = a.copy()
id(a) == id(a_2), id(a) == id(a_copy)
```

```python
a = [[1, 2, 3]] * 5
print(a)
a[1][1] = 100
a[0][2] = True
a
```

```python
a = [[1, 2, 3] for _ in range(5)]
print(a)
a[1][1] = 100
a[0][2] = True
a
```
