```python
def outer_fun(text):
    def inner_fun(symb):
        return f'{text}: {symb:^10}'
    return inner_fun('abc')
outer_fun('hey')
```

```python
def my_decorator(func):
    def wrapper():
        print('Start')
        func()
        print('End')
    return wrapper

@my_decorator
def say_hello():
    print('\thello')

# my_decorator(say_hello)()
say_hello()
```

```python
from functools import wraps

def foo():
    ...

def my_decor(f):
    @wraps(f)
    def wrapper():
        ...
    return wrapper

@my_decor
def boo():
    ...

foo.__name__, boo.__name__
```

```python
def check_zero(f):
    def wrapper(n1, n2):
        if n2 == 0:
            return None
        return f(n1, n2)
    return wrapper

@check_zero
def div(a, b):
    return a / b

print(div(div(4, 2), div(0, 10)))
```

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)
factorial(5)
```
