```python
def area (length: int | float, width: int | float) -> int | float | str:
    """Finding are of the right angled triangle."""
    return 1 / 2 * length * width
    # return 'area is ,,,'
```

```python
area(3, 4)
```

```python
def area(length: int | float, width: int | float) -> float:
    """Calculate area

    >>> area(2, 3)
    6.0
    >>> area(2.5, 2.5)
    6.25
    """
    return length * width
```

```python
x = area('2', 3)
print(x + 2)
```
