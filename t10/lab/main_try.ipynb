```python
def loop_generator(n):
    sum = 0
    for i in range(n):
        sum += i
        yield sum

n = 10
# n = 1000000000000000 # would throw memory error if it was retrn instead of yield
sum_n = loop_generator(n)

while True:
    try:
        input('Pause')
        print(sum_n.__next__())
    except:
        print('we hit the limit')
        break
```

```python
def reverse_letter(word):
    return ''.join(filter(lambda x: x.isalpha(), word[::-1]))
reverse_letter('krishan'), reverse_letter('ultr53o?n')
```

```python
def diagonal_sum(array):
    return sum(map(lambda sub_arr: sub_arr[1][sub_arr[0]], list(enumerate(array))))
diagonal_sum([[1, 2, 3], [4, 5, 6], [7, 8, 9]])
```
