### Iterator

```python
iter(range(10)), iter([1, 2, 3]), iter('123'), iter({'a': 1, 'b': 2, 'c': 3})
```

```python
next(iter({'a': 1, 'b': 2, 'c': 3})) # keys
```

```python
it = iter([1, 2, 3])
next(it), next(it), list(it), list(it)
```

```python
it = iter([1, 2])
next(it)
next(it)
next(it)
```

```python
it = iter([1, 2, 3])
print(max(it)) # looked over all elements
min(it)
```

```python
it = iter([1, 2, 3, 4, 5])
print(3 in it) # 1 no, 2 no, 3 yes, stop
list(it)
```

```python
s = [1, 2, 3, 4]
it = iter(s.pop, 3)
print(next(it))
next(it)
```

```python
s = 'abc123'
it = iter(s.isalpha, True)
next(it) # 'abc123'.isalpha
```

```python
it = iter([15, 10, 1, 3])
next(it), max(it), sum(it)
```

```python
s = [15, 10, 1, 3]
list(iter(s.pop, 10)) # 10 -> StopIteration
```

### Generator

```python
def generate(n):
    for i in range(1, n, 2):
        yield i
        # = return i

s = generate(10)
print(type(s))

print(next(s))
print(next(s))
print(min(s))
```

```python
def generate(n):
    for i in range(1, n, 2):
        yield i
        # = return i

s = generate(5)
list(s), list(s)
```

```python
def generate(n):
    for i in range(1, n, 2):
        yield i
        # = return i

s1 = generate(5)
s2 = (i for i in range(1,5, 2))
s1, s2, list(s1) == list(s2)
```

```python
def some_seq(n):
    a, b = 0, 1
    while a < n:
        yield a
        yield b
        a, b = 6, 7
it = some_seq(10)
next(it), next(it), next(it), next(it)
```

```python
def generate(n):
    for i in range(1, n, 2):
        yield i
        return i
s = generate(10)
print(list(s)) # before return

s = generate(10)
print(next(s))
print(next(s)) # approaches return
```
